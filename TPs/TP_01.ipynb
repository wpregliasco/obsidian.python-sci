## Carga de datos

```python
# archivo
f=open('SAZS_2020_2026.csv', 'rt')
data = f.read()
f.close()

print(len(data), type(data))
print('-'*70)
print(data[:700])
```

```python
# Separación en líneas
data_lines = data.split('\n')
print(len(data_lines), type(data_lines))
print('-'*70)
print(data_lines[0])
print(data_lines[1])
```

```python
# Lectura del header por separado
headers = data_lines[0].split(',')
print(len(headers), type(headers))
print('-'*70)
print(headers)
```

```python
# Tiempos, uso timestampas que son segundo desde 1970-01-01
#
timestamp = []
for row in range(1, len(data_lines)):
    timestamp.append(data_lines[row].split(',')[2])


for item in timestamp[:10]:
    print(item)
```

```python
# Temperatura
#
temp = []
for row in range(1, len(data_lines)):
    temp.append(data_lines[row].split(',')[6])

for item in temp[:10]:
    print(item)
```

```python
# Persión atmosférica
#
press = []
for row in range(1, len(data_lines)):
    press.append(data_lines[row].split(',')[10])

for item in press[:10]:
    print(item)            
```
