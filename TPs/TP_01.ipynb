## Carga de datos

```python
# archivo
f=open('SAZS_2020_2026.csv', 'rt')
data = f.read()
f.close()

print(len(data), type(data))
print('-'*70)
print(data[:700])
```

```python
# Separación en líneas
data_lines = data.split('\n')
print(len(data_lines), type(data_lines))
print('-'*70)
print(data_lines[0])
print(data_lines[1])
```

```python
# Lectura del header por separado
headers = data_lines[0].split(',')
print(len(headers), type(headers))
print('-'*70)
print(headers)
```

```python
# Tiempos
times = []
for row in range(1, len(data_lines)):
    if data_lines[row] != '':
        times.append(data_lines[row].split(',')[1])

print(times[:50])
```

```python
# Metar
metar = []
for row in range(1, len(data_lines)):
    if data_lines[row] != '':
        metar.append(data_lines[row].split(',')[5])

for item in metar[:10]:
    print(item)
```

```python
# Persión atmosférica
press = []
for row in metar:
    found = False
    for item in row.split():
        if item.startswith('Q'):
            press.append(item)
            found = True
    if found == False:
        press.append(None)

for item in press[:10]:
    print(item)            
```
