```python
from matplotlib import pyplot as plt
import numpy as np

N = 360
x = np.linspace(0.0001,6*np.pi,N)
y = sin(x)/x

plt.plot(x,y,'r-')
plt.grid()
plt.show()
```

Esto es la ecuación:
$$ y = \frac{\sin x}{x}  \ \ \ \ \ x \in [0,6\pi]
$$
