## 1. Separacion de pares e impares

Escriba una función separa_pares_impares(lista) que reciba una lista de enteros y devuelva dos listas: una con los números pares y otra con los impares, conservando el orden original.

Especificaciones:
 * La función debe devolver una tupla (pares, impares).
 * No debe modificar la lista original.

Ejemplo de uso:

```python
pares, impares = separa_pares_impares([1, 2, 3, 4, 5, 6])
print(pares)   # [2, 4, 6]
print(impares) # [1, 3, 5]
```
